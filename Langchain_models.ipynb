### Embedding Model using Closed Model

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
from huggingface_hub import HfApi, InferenceClient
from google.colab import userdata

HF_TOKEN = userdata.get("HF_TOKEN")

api = HfApi()

# Search for feature-extraction / embedding models
models = api.list_models(
    filter="feature-extraction",
    limit=20
)

client = InferenceClient(
    provider="auto",
    api_key=HF_TOKEN
)

for m in models:
    print(f"Testing: {m.id}")

    try:
        embedding = client.feature_extraction(
            text="What is the full form of LLM?",
            model=m.id
        )

        print(f"✅ Supported: {m.id}")
        print(f"Embedding dimension: {len(embedding)}")
        break

    except Exception as e:
        print(f"❌ Not supported: {m.id}")
```

```python
embedding_models = [
    "sentence-transformers/all-MiniLM-L6-v2",
    "BAAI/bge-small-en-v1.5",
    "BAAI/bge-base-en-v1.5",
    "intfloat/e5-small-v2",
]

for model_name in embedding_models:
    try:
        embedding = client.feature_extraction(
            text="Hello world",
            model=model_name
        )

        print(f"✅ Works: {model_name}")
        print(f"Dimensions: {len(embedding)}")

    except Exception as e:
        print(f"❌ Failed: {model_name}")
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

text = "What is the full form of LLM?"
vector = embeddings.embed_query(text)
```

```python
print(str(vector))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

text1 = "What is AI?"
text2 = "Explain Artificial Intelligence"

v1 = embeddings.embed_query(text1)
v2 = embeddings.embed_query(text2)

score = cosine_similarity([v1], [v2])

print(score)
```

```python
docs = [
    "LangChain is a framework for LLM applications.",
    "Vector databases store embeddings.",
    "FAISS is an open-source vector store."
]

vectors = embeddings.embed_documents(docs)

print(len(vectors))
print(len(vectors[0]))
```

#### Semantic Search

```python
from sklearn.metrics.pairwise import cosine_similarity
documents= [
    "Virat Kohli is an Indian cricketer known for his aggressive batting and leadership.",
    "MS Dhoni is a former Indian captain famous for his calm demeanor and finishing skills.",
    "Sachin Tendulkar, also known as the 'God of Cricket', holds many batting records.",
    "Rohit Sharma is known for his elegant batting and record-breaking double centuries.",
    "Jasprit Bumrah is an Indian fast bowler known for his unorthodox action and yorkers."
]
query = "Tell me about Bumrah"

document_embedding = embeddings.embed_documents(documents)
query_embedding = embeddings.embed_query(query)

scores = cosine_similarity([query_embedding], document_embedding)[0]
sorted_scores = sorted(list(enumerate(scores)),key = lambda x: x[1], reverse=True)
print(sorted_scores)
```

```python
index,score = sorted_scores[0]
print(f"Query: {query}")
print(f"Document: {documents[index]}")
print(f"Score: {score}")
```
